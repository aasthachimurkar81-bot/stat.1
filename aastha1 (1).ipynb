### Q9: Mean and Standard Deviation in Real Life

Monthly electricity usage:

$$

|     Usage | Difference from Mean | Squared Difference |

## Q10: Basketball Player’s Scoring Performance

Points in 8 games:

**[15, 18, 20, 22, 25, 17, 19, 21]**

### 1. Mean

$$
\text{Mean}=\frac{15+18+20+22+25+17+19+21}{8}
$$

$$
=\frac{157}{8}=\boxed{19.625}
$$

**Mean = 19.625 points**

---

### 2. Median

First arrange the values in ascending order:

**15, 17, 18, 19, 20, 21, 22, 25**

There are 8 values, so the median is the average of the 4th and 5th values:

$$
\text{Median}=\frac{19+20}{2}=\boxed{19.5}
$$

**Median = 19.5 points**

---

### 3. Mode

No number appears more than once.

**Mode = No mode**

---

### 4. Range

$$
\text{Range}=\text{Maximum}-\text{Minimum}
$$

$$
=25-15=\boxed{10}
$$

**Range = 10 points**

---

### 5. Standard Deviation

Using the population standard deviation formula:

$$
\sigma=\sqrt{\frac{\sum(x-\bar{x})^2}{n}}
$$

Mean = 19.625.

The sum of squared deviations is:

$$
\sum(x-\bar{x})^2=67.875
$$

Therefore,

$$
\sigma=\sqrt{\frac{67.875}{8}}
$$

$$
=\sqrt{8.484375}\approx\boxed{2.91}
$$

**Standard deviation ≈ 2.91 points**

### Final Answers

| Measure                |            Answer |
| ---------------------- | ----------------: |
| **Mean**               | **19.625 points** |
| **Median**             |   **19.5 points** |
| **Mode**               |       **No mode** |
| **Range**              |     **10 points** |
| **Standard Deviation** |   **2.91 points** |

### Insights

These measures tell us that:

* The player scores about **19.6 points per game on average**.
* The mean and median are very close, showing that the scores are **fairly balanced around 20 points**.
* There is **no mode** because every score is different.
* The range of **10 points** shows the difference between the highest and lowest scores.
* The standard deviation of about **2.91 points** indicates that the scores generally stay **fairly close to the average**, so the player's scoring performance is relatively consistent.
